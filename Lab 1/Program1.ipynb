Data Preprocessing (Medium)

*   Load the MNIST handwritten digit dataset and perform the following
*   Normalize the pixel values of the images.
*   Apply one-hot encoding to the target labels.
*   Split the data into training, validation, and test sets.
*   Dataset: https://github.com/iamavieira/handwritten-digits-mnist

```python
import numpy as np
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
print("Top 5 labels:")
print(y_train[:5])

print("\nTop 5 images (pixel values):")
print(X_train[:5])
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
X_train, X_val, y_train, y_val = train_test_split( #split training
    X_train, y_train, test_size=0.2, random_state=42
)
```

```python
print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Test:", X_test.shape, y_test.shape)
```
