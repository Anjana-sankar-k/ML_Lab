```python
import pandas as pd
import numpy as np

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter']
]

print("Sample Transactions:")
for t in transactions:
    print(t)
```

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_array, columns=te.columns_)

print("One-Hot Encoded Transaction Data:")
df.head()
```

```python
item_support = df.mean().sort_values(ascending=False)

print("Support for Individual Items:")
print(item_support)
```

```python
min_support_threshold = 0.3  # You can change this

frequent_itemsets = apriori(df,
                            min_support=min_support_threshold,
                            use_colnames=True)

frequent_itemsets = frequent_itemsets.sort_values(by="support", ascending=False)

print("Frequent Itemsets:")
frequent_itemsets
```

```python
min_confidence_threshold = 0.6

rules = association_rules(frequent_itemsets,
                          metric="confidence",
                          min_threshold=min_confidence_threshold)

rules = rules.sort_values(by="confidence", ascending=False)

print("Association Rules:")
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

```python
strong_rules = rules[(rules['confidence'] >= 0.6) &
                     (rules['lift'] > 1)]

print("Strong Association Rules:")
strong_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```
